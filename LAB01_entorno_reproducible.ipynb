<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Un entorno aislado es un directorio con su propia copia del intérprete y de las bibliotecas. Sirve para dos cosas: que dos proyectos con versiones incompatibles convivan en la misma máquina, y que las dependencias del proyecto queden <strong>escritas</strong> en lugar de vivir en la memoria de quien lo instaló.</p>
<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Las tres modalidades admitidas en la asignatura son equivalentes. Elegí una.</p>
<ul style="margin: 6px 0 8px 18px; color: #222; font-size: 0.97em; line-height: 1.6; padding-left: 14px;">
<li><strong>Entorno virtual local</strong> (<code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">venv</code>): máquina propia, trabajo cotidiano. Costo bajo, depende del Python instalado.</li>
<li><strong>Contenedor</strong>: reproducibilidad estricta, o sistema operativo distinto al del resto del equipo. Costo medio, aísla también el sistema.</li>
<li><strong>Servicio de cuadernos en la nube</strong>: máquina limitada, o clase sin instalación previa. Costo nulo al inicio, pero hay que reinstalar en cada sesión.</li>
</ul></div>

<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Por eso el techo. Una dependencia sin techo no es una dependencia declarada: es una apuesta a que nadie publique una versión nueva antes de la entrega.</p></div>

````python
# --- Qué versiones tengo realmente instaladas --------------------------------
# Se imprime con un ciclo y no con una tabla: las tablas llegan con la
# manipulación de datos, que es el tema de la unidad Datos y análisis exploratorio.
from importlib.metadata import version, PackageNotFoundError

paquetes = ['numpy', 'pandas', 'scikit-learn', 'matplotlib', 'seaborn',
            'plotly', 'xgboost', 'mlxtend', 'statsmodels']

faltan = []
for nombre in paquetes:
    try:
        print(f'{nombre:<14} {version(nombre)}')
    except PackageNotFoundError:
        print(f'{nombre:<14} FALTA')
        faltan.append(nombre)

print()
if faltan:
    print('Faltan paquetes. Ejecutar:  pip install -r requirements.txt')
else:
    print('Todas las dependencias declaradas están instaladas.')
````

<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;"><pre style='background:#f7fafd;border-radius:6px;padding:10px 14px;font-size:0.88em;line-height:1.5;margin:0 0 10px 0;'>proyecto/

<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">La regla que más se incumple es la primera. Un archivo de <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">datos/crudos/</code> no se edita jamás. Toda corrección se hace por código, de modo que quede registrada y se pueda deshacer. Si se corrige a mano, dentro de tres semanas nadie puede decir qué se cambió ni por qué, y el análisis deja de ser auditable.</p>
<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">La segunda en importancia: todo lo que está en <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">derivados/</code> y en <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">salidas/</code> tiene que poder borrarse y regenerarse con un comando. Si algo se rompe al borrarlo, es que ese archivo era en realidad un dato crudo disfrazado.</p></div>

<div style="background: #e8f4fd; border-left: 5px solid #0055a5; border-radius: 8px; padding: 16px 20px; margin: 6px 0; box-shadow: 0 2px 6px rgba(0,0,0,0.07);"><div style="margin-bottom: 8px;"><span style="background: #0055a5; color: white; font-weight: bold; font-size: 0.95em; padding: 2px 14px; border-radius: 20px;">🔵 Ejercicio 1</span><b style="color: #003366; font-size: 1.0em; margin-left: 10px;">Armar la estructura del equipo</b></div>
<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Completá la celda para que cree el árbol de directorios del proyecto dentro de <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">PROYECTO</code>, y para que deje un archivo <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">.gitkeep</code> en las tres carpetas que arrancan vacías.</p>
<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Git no versiona directorios vacíos: sin ese archivo, quien clone el repositorio no recibe las carpetas y el primer <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">cargar()</code> le falla.</p></div>

<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Un entorno es reproducible cuando lo que dice el archivo de dependencias es lo que hay instalado. Escribí la función que lee ese archivo y devuelve la lista de nombres de paquete que declara.</p>
<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">De la línea <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">pandas&gt;=2.1,&lt;3.1  # núcleo</code> tiene que salir <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">pandas</code>. Las líneas vacías y las que son sólo comentario no cuentan.</p></div>

````python
import re


def declarados(texto):
    """Nombres de paquete que declara un archivo de dependencias."""
    nombres = []
    for linea in texto.splitlines():
        # Sacá el comentario, limpiá los espacios y salteá lo que quede vacío.
        linea = linea.split('#')[0].strip()
        if not linea:
            continue
        
        # Quedate con el nombre, o sea lo que hay antes del primer
        # signo de comparación. `re.split` con esos signos lo resuelve en una línea.
        nombre = re.split(r'[<>=!~]', linea)[0].strip()
        if nombre:
            nombres.append(nombre)
    return nombres


ARCHIVO = RAIZ / "requirements.txt"
pedidos = declarados(ARCHIVO.read_text(encoding="utf-8"))
print(len(pedidos), "paquetes declarados")

from importlib.metadata import version, PackageNotFoundError
for nombre in pedidos:
    try:
        print(" ", nombre, version(nombre))
    except PackageNotFoundError:
        print(" ", nombre, "NO INSTALADO")
````

````python
ejemplo = """# comentario suelto
pandas>=2.1,<3.1  # núcleo

scikit-learn==1.7.0
pyarrow
   seaborn>=0.13
"""
assert declarados(ejemplo) == ["pandas", "scikit-learn", "pyarrow", "seaborn"], \
    declarados(ejemplo)

# el archivo real del repositorio declara varios y ningún nombre queda vacío
assert len(pedidos) > 5, 'el archivo real del repositorio declara varios y ningún nombre queda vacío'
assert all(nombre != "" for nombre in pedidos), 'el archivo real del repositorio declara varios y ningún nombre queda vacío'
print("Ejercicio 2 verificado ✔")
````

<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Fija la aleatoriedad, no el resultado numérico exacto. Cambios de versión de la biblioteca, de arquitectura del procesador o de cantidad de hilos pueden mover los últimos decimales.</p>
<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Por eso la reproducibilidad se apoya en tres patas y no en una: <strong>semilla fija, versiones acotadas y datos verificados</strong>. Una entrega cuyo resultado cambia entre ejecuciones se devuelve sin corregir, porque no hay forma de verificar lo que afirma.</p></div>

<div style="background: #e8f4fd; border-left: 5px solid #0055a5; border-radius: 8px; padding: 16px 20px; margin: 6px 0; box-shadow: 0 2px 6px rgba(0,0,0,0.07);"><div style="margin-bottom: 8px;"><span style="background: #0055a5; color: white; font-weight: bold; font-size: 0.95em; padding: 2px 14px; border-radius: 20px;">🔵 Ejercicio 2</span><b style="color: #003366; font-size: 1.0em; margin-left: 10px;">Un sorteo que se puede repetir</b></div>
<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Escribí una función que elija al azar una cantidad de elementos de una lista y que, con la misma semilla, devuelva siempre la misma selección.</p>
<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Es el mecanismo que sostiene todo el cuatrimestre: el mismo sorteo con la misma semilla da el mismo resultado en cualquier máquina. Usá el módulo <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">random</code> de la biblioteca estándar, que es el que se mostró arriba.</p></div>

````python
import random


def sortear(elementos, cantidad, semilla):
    """Devuelve `cantidad` elementos de `elementos`, siempre los mismos
    para la misma semilla."""
    # Creá un generador propio con la semilla recibida y usalo
    # para tomar la muestra. No toques el generador global.
    generador = random.Random(semilla)
    return generador.sample(elementos, cantidad)


equipos = [f'equipo {i}' for i in range(1, 13)]
print('con semilla 42 :', sortear(equipos, 4, 42))
print('otra vez con 42:', sortear(equipos, 4, 42))
print('con semilla 7  :', sortear(equipos, 4, 7))
````

<div style="background: #e9f7ec; border-left: 5px solid #2e7d32; border-radius: 8px; padding: 16px 20px; margin: 6px 0; box-shadow: 0 2px 6px rgba(0,0,0,0.07);"><div style="margin-bottom: 8px;"><span style="background: #2e7d32; color: white; font-weight: bold; font-size: 0.95em; padding: 2px 14px; border-radius: 20px;">✅ Verificación</span></div>
<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Ejecutá esta celda: si no lanza error, el ejercicio está resuelto.</p></div>

````python
# la misma semilla da la misma selección
assert sortear(equipos, 4, 42) == sortear(equipos, 4, 42), 'la misma semilla da la misma selección'
# semillas distintas dan selecciones distintas
assert sortear(equipos, 4, 42) != sortear(equipos, 4, 7), 'semillas distintas dan selecciones distintas'
# devuelve la cantidad pedida, sin repetir, y todos salen de la lista
m = sortear(equipos, 5, 2026)
assert len(m) == 5 and len(set(m)) == 5, 'devuelve la cantidad pedida, sin repetir, y todos salen de la lista'
assert all(x in equipos for x in m), 'devuelve la cantidad pedida, sin repetir, y todos salen de la lista'
# y NO toca el generador global: dos llamadas seguidas no mueven random.random()
random.seed(1); antes = random.random()
random.seed(1); sortear(equipos, 3, 99); despues = random.random()
assert antes == despues, 'usá random.Random(semilla), no random.seed(semilla)'
print("Ejercicio 3 verificado ✔")
````

<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Escribí una función que reciba el texto de una celda de código y devuelva la lista de faltas contra las reglas de esta clase. Son tres, y son las que hicieron que trece cuadernos de la versión anterior de esta asignatura hoy no corran en ninguna máquina.</p>
<ul style="margin: 6px 0 8px 18px; color: #222; font-size: 0.97em; line-height: 1.6; padding-left: 14px;">
<li><strong>ruta absoluta</strong>: la línea trae una ruta que empieza en la raíz del disco.</li>
<li><strong>URL</strong>: la línea trae <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">http://</code> o <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">https://</code>.</li>
<li><strong>azar sin semilla</strong>: la línea sortea algo y en ninguna línea anterior se fijó la semilla.</li>
<li>Las líneas que empiezan con <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">#</code> y las vacías no se revisan.</li>
</ul></div>

````python
# ejemplo deliberado de lo que NO hay que hacer
def revisar(celda):
    """Faltas contra las reglas de reproducibilidad, una por línea."""
    faltas = []
    hay_semilla = False
    for numero, linea in enumerate(celda.splitlines(), start=1):
        limpia = linea.strip()
        if limpia == "" or limpia.startswith("#"):
            continue
        # 1. Si la línea fija la semilla, anotalo y pasá a la siguiente.
        if "random.seed" in limpia or "np.random.seed" in limpia:
            hay_semilla = True
            continue
            
        # 2. Agregá a `faltas` una tupla (numero, motivo) por cada falta.
        
        # - Ruta absoluta: detecta ventanas (C:\ o C:/) y unix ("/home" o '/var')
        if re.search(r"[a-zA-Z]:[\\/]", limpia) or re.search(r"['\"]/[a-zA-Z0-9_]", limpia):
            faltas.append((numero, "ruta absoluta"))
            
        # - URL: detecta enlaces web
        if "http://" in limpia or "https://" in limpia:
            faltas.append((numero, "URL"))
            
        # - Azar sin semilla: usa el módulo random pero no es un import y la semilla no está fija
        if not hay_semilla and "random." in limpia and "import " not in limpia:
            faltas.append((numero, "azar sin semilla"))
    return faltas


EJEMPLO = """# lectura del conjunto
import random
datos = open("C:\\Users\\ana\\ventas.csv").read()
tabla = descargar("https://docs.google.com/spreadsheets/d/1a2b3c")
muestra = random.sample(range(100), 10)
"""

for numero, motivo in revisar(EJEMPLO):
    print("línea", numero, "->", motivo)
````

<p style="margin: 0 0 8px 0; color: #222; font-size: 0.97em; line-height: 1.6;">Ejecutá esta celda: si no lanza error, el ejercicio está resuelto.</p></div>

````python
hallado = revisar(EJEMPLO)
motivos = sorted(motivo for numero, motivo in hallado)
assert motivos == ["URL", "azar sin semilla", "ruta absoluta"], motivos

# el comentario de la primera línea no se revisa
assert all(numero != 1 for numero, motivo in hallado), 'el comentario de la primera línea no se revisa'

# con la semilla fijada antes, el sorteo deja de ser una falta
con_semilla = """import random
random.seed(42)
muestra = random.sample(range(9), 3)
"""
assert revisar(con_semilla) == [], 'con la semilla fijada antes, el sorteo deja de ser una falta'

# y una celda limpia no devuelve nada
assert revisar("total = sum([1, 2, 3])") == [], 'y una celda limpia no devuelve nada'
print("Ejercicio 4 verificado ✔")
````

<li>Limpiar las salidas antes de confirmar: <code style="background: #dbeafe; color: #003366; padding: 1px 5px; border-radius: 4px; font-size: 0.92em;">jupyter nbconvert --clear-output --inplace cuaderno.ipynb</code>.</li>
